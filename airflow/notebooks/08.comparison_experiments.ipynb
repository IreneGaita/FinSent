```
# evaluate_models.ipynb

# 📦 Import librerie
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 📊 Caricamento del CSV con risultati
file_path = "../data/reports/comparison_table.csv"  # sistema il path se necessario
df = pd.read_csv(file_path)

# 🔍 Panoramica dei dati
print("Shape del DataFrame:", df.shape)
df.head()
```

```
# 1️⃣ Ordinamento per accuracy e primi esperimenti
df_sorted = df.sort_values(by="accuracy", ascending=False).reset_index(drop=True)
df_sorted.head(5)
```

```
# 2️⃣ Visualizzazione barplot - Accuracy
plt.figure(figsize=(10,6))
sns.barplot(data=df_sorted, x="experiment", y="accuracy", hue="embedding")
plt.xticks(rotation=45, ha="right")
plt.title("Accuracy per Esperimento")
plt.tight_layout()
plt.show()

```

```
# 3️⃣ Visualizzazione F1-score per classe
df_melt = df_sorted.melt(
    id_vars=["experiment", "embedding"],
    value_vars=["f1_negative", "f1_neutral", "f1_positive"],
    var_name="class", value_name="f1_score"
)

plt.figure(figsize=(12,6))
sns.barplot(data=df_melt, x="class", y="f1_score", hue="embedding")
plt.title("F1-score per Classe e Embedding")
plt.tight_layout()
plt.show()
```

```
# 4️⃣ Analisi bilanciamento (balancing vs no balancing)
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x="balancing", y="f1_negative")
plt.title("Effetto del Bilanciamento sulla F1-score della classe 'negative'")
plt.tight_layout()
plt.show()
```

```
# 5️⃣ Top 3 modelli definiti
top3 = df_sorted.head(3)
print("\n Top 3 esperimenti:")
print(top3[["experiment", "embedding", "balancing", "accuracy", "f1_negative", "f1_positive", "f1_neutral"]])

```

```
# 6️⃣ Identificazione automatico del best model
best = df_sorted.iloc[0]
print(f"\n🎯 Modello migliore: {best['experiment']}")
print(best.to_dict())
```
